# Interpreting the Culmen Pair Plot

Culmen length and depth together are enough to tell the three penguin species apart, though a few points overlap at the edges.  
Adelie penguins have a short but deep culmen, Gentoo penguins have a long but shallow culmen, and Chinstrap penguins have a long and deep culmen.  
Neither measurement works alone: length can't separate Gentoo from Chinstrap, and depth can't separate Adelie from Chinstrap.  
But in the scatter plot, the two measurements combine to form three distinct clusters.
